```python
df = pd.read_csv("C:\\Users\\pawan\\Downloads\\EnvironmentProject\\ML\\annual-freshwater-withdrawals.csv")
remove_entities = ["World","Asia (excl. China and India)","Europe","Low-income countries","Europe (excl. EU-27)",
                   "Europe (excl. EU-28)","European Union (27)","Lower-middle-income countries","North America",
                    "North America (excl. USA)","European Union (28)","High-income countries","Upper-middle-income countries","Asia",
                  "Europe and Central Asia (WB)","East Asia and Pacific (WB)","European Union (27)"," Afghanistan and Pakistan (WB)","South Asia (WB)"]
df = df[~df["Entity"].isin(remove_entities)]
#df["Entity"].unique()
X = df[["Entity","Year"]]
y = df["Annual freshwater withdrawals"]
train = df[df["Year"] <= 2018]
test = df[df["Year"] > 2018]
X_train = train[["Entity", "Year"]]
y_train = train["Annual freshwater withdrawals"]

X_test = test[["Entity", "Year"]]
y_test = test["Annual freshwater withdrawals"]
preprocessor = ColumnTransformer(
    transformers=[("country",OneHotEncoder(handle_unknown="ignore"), ["Entity"])],
    remainder="passthrough"
)
Water_model = Pipeline(steps=[("preprocessor", preprocessor),("regressor", RandomForestRegressor(random_state=42))]
)
Water_model.fit(X_train, y_train)
predictions = Water_model.predict(X_test)
print("Actual minimum:", y_test.min())
print("Actual maximum:", y_test.max())
print("Actual mean:", y_test.mean())

print("MAE:", mean_absolute_error(y_test, predictions))
mean_actual = y_test.mean()
print("MAE %:", (mae / mean_actual) * 100)
print("R²:", r2_score(y_test, predictions))
```

```python
## AIR POLLUTION
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
df = pd.read_csv("C:\\Users\\pawan\\Downloads\\EnvironmentProject\\ML\\outdoor-air-pollution-exposure.csv")
target = "Outdoor air pollution exposure (population-weighted PM2.5)"

train = df[df["Year"] <= 2018]
test = df[df["Year"] > 2018]
countries = [
    "Australia",
    "Bangladesh",
    "Brazil",
    "China",
    "Egypt",
    "Germany",
    "India",
    "Japan",
    "Kenya",
    "Kuwait",
    "Pakistan",
    "Sweden",
    "Uganda",
    "United Kingdom",
    "United States"
]
linear_models = {}

for country in countries:

    country_train = train[train["Entity"] == country]

    X = country_train[["Year"]]
    y = country_train[target]

    model = LinearRegression()

    model.fit(X, y)

    linear_models[country] = model

print("All country models trained successfully.")
test_predictions = []

for country in countries:

    country_test = test[test["Entity"] == country]

    if len(country_test) == 0:
        continue

    model = linear_models[country]

    predictions = model.predict(country_test[["Year"]])

    for actual, predicted in zip(
        country_test[target],
        predictions
    ):
        test_predictions.append({
            "Entity": country,
            "Actual": actual,
            "Predicted": predicted
        })
results = pd.DataFrame(test_predictions)
print(results.head())
mae = mean_absolute_error(
    results["Actual"],
    results["Predicted"]
)

r2 = r2_score(
    results["Actual"],
    results["Predicted"]
)

print("Linear Regression MAE :", mae)
print("Linear Regression R² :", r2)        
```

```python
import json
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Updated Function (added allow_negative=False)
def train_and_export(df, entity_col, year_col, value_col, target_countries, future_years, output_file, min_year=2005, allow_negative=False):
    output = {}
    for country in target_countries:
        subset = df[df[entity_col] == country].sort_values(year_col)
        subset = subset[subset[year_col] >= min_year]

        if len(subset) < 5:
            continue

        X_train = subset[[year_col]]
        y_train = subset[value_col]

        model = LinearRegression()
        model.fit(X_train, y_train)

        X_future = pd.DataFrame({year_col: future_years})
        
        # Only clip to 0 if allow_negative is False
        preds = model.predict(X_future)
        if not allow_negative:
            preds = preds.clip(min=0)

        output[country] = [
            {"country": country, "year": int(y), "value": round(float(v), 2), "predicted": True}
            for y, v in zip(future_years, preds)
        ]

    with open(output_file, "w") as f:
        json.dump(output, f, indent=2)
    print(f"Exported {output_file}")

# 2. Make sure your variables are defined
target_countries = [
    "Australia", "Bangladesh", "Brazil", "China", "Egypt", 
    "Germany", "India", "Japan", "Kenya", "Kuwait", 
    "Pakistan", "Sweden", "Uganda", "United Kingdom", "United States"
]
future_years = [2025, 2026, 2027, 2028, 2029, 2030]

# (Assume co2_df, pm25_df, water_df, forest_df are already loaded via pd.read_csv here)
co2_df = pd.read_csv("C:\\Users\\pawan\\Downloads\\EnvironmentProject\\ML\\co2-emissions-per-capita.csv")
pm25_df = pd.read_csv("C:\\Users\\pawan\\Downloads\\EnvironmentProject\\ML\\outdoor-air-pollution-exposure.csv") # Apni PM2.5 file ka naam dalein
water_df = pd.read_csv("C:\\Users\\pawan\\Downloads\\EnvironmentProject\\ML\\annual-freshwater-withdrawals.csv") # Apni Water file ka naam dalein
forest_df = pd.read_csv("C:\\Users\\pawan\\Downloads\\EnvironmentProject\\ML\\annual-change-forest-area.csv") # Apni Forest file ka naam dalein
# 3. Call the functions (Note the True on the last line!)
train_and_export(co2_df, "Entity", "Year", "CO₂ emissions per capita", target_countries, future_years, "co2_predictions.json")
train_and_export(pm25_df, "Entity", "Year", "Outdoor air pollution exposure (population-weighted PM2.5)", target_countries, future_years, "pm25_predictions.json")
train_and_export(water_df, "Entity", "Year", "Annual freshwater withdrawals", target_countries, future_years, "water_predictions.json")

# allow_negative=True allows this specific model to predict negative forest change (deforestation)
train_and_export(forest_df, "Entity", "Year", "Annual change in forest area", target_countries, future_years, "forest_predictions.json", allow_negative=True)
```
